```python
airport_stats = duckdb.sql("""SELECT carrier_name, SUM(arr_del15) / SUM(arr_flights) as percent_delayed
                       FROM df
                       GROUP BY carrier_name
                       ORDER BY percent_delayed""").df()

print(airport_stats)
```

```python
airport_stats2 = duckdb.sql("""SELECT carrier_name, SUM(arr_del15) / SUM(arr_flights) as percent_delayed, SUM(carrier_ct) / (SUM(carrier_ct) + SUM(weather_ct) + SUM(nas_ct) + SUM(security_ct) + SUM(late_aircraft_ct)) as carrier_pt, SUM(weather_ct) / (SUM(carrier_ct) + SUM(weather_ct) + SUM(nas_ct) + SUM(security_ct) + SUM(late_aircraft_ct)) as weather_pt, SUM(nas_ct) / (SUM(carrier_ct) + SUM(weather_ct) + SUM(nas_ct) + SUM(security_ct) + SUM(late_aircraft_ct)) as nas_pt, SUM(security_ct) / (SUM(carrier_ct) + SUM(weather_ct) + SUM(nas_ct) + SUM(security_ct) + SUM(late_aircraft_ct)) as security_pt, SUM(late_aircraft_ct) / (SUM(carrier_ct) + SUM(weather_ct) + SUM(nas_ct) + SUM(security_ct) + SUM(late_aircraft_ct)) as late_aircraft_pt
                       FROM df
                       GROUP BY carrier_name
                       ORDER BY percent_delayed""").df()

print(airport_stats2)
```

```python
airport_stats3 = duckdb.sql("""SELECT airport, SUM(arr_del15) / SUM(arr_flights) as percent_delayed, SUM(arr_del15), SUM (arr_flights) AS total_flights
                       FROM df
                       GROUP BY airport
                       ORDER BY total_flights""").df()

airport_stats3.tail(15)
```
